```python
from pyspark.sql.types import IntegerType, StringType, StructField, StructType

# Occupation IDs map to this list (index = occupation_id) from the MovieLens dataset
occupation_list = [
    "other",                    # 0
    "academic/educator",        # 1
    "artist",                   # 2
    "clerical/admin",           # 3
    "college/grad student",     # 4
    "customer service",        # 5
    "doctor/health care",       # 6
    "executive/managerial",     # 7
    "farmer",                   # 8
    "homemaker",                # 9
    "K-12 student",             # 10
    "lawyer",                   # 11
    "programmer",               # 12
    "retired",                  # 13
    "sales/marketing",          # 14
    "scientist",                 # 15
    "self-employed",            # 16
    "technician/engineer",      # 17
    "tradesman/craftsman",      # 18
    "unemployed",               # 19
    "writer"                    # 20
]

occupations_schema = StructType([
    StructField("occupation_id", IntegerType(), False),
    StructField("occupation_name", StringType(), False)
])

occupations_data = [(i, name) for i, name in enumerate(occupation_list)]

occupations_df = spark.createDataFrame(occupations_data, schema=occupations_schema)

occupations_table = (
    occupations_df.write
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("workspace.default.occupations")
)

display(occupations_df)
```
