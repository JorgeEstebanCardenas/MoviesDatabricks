```python
movie_df = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("mergeSchema", "true")
    .load("/Workspace/Users/jorgestebance@gmail.com/MoviesDatabricks/data/archive/movies.csv")
)

movie_table = (
    movie_df.write
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("workspace.default.movies")

)
```
